```python
# Create Dataframe
data={"Name":["A","B","C"],
      "Marks":[85,90,78]}
df=pd.DataFrame(data)
df
```

```python
df.head()
df.tail()
```

```python
# Find the structure of the DataFrame
print(df.shape)
print(df.ndim)
print(df.size)
# Display Column names
print(df.columns)
```

```python
# Check Data Types of Each Column
print(df.dtypes)
```

```python
# Acess Individual Columns
df["Name"]
```

```python
# Access Name of first row
print(df.loc[0,"Name"])
print(df.iloc[2,1])
```

```python
# Add a New Column
df["Grade"]=["First Class","Distinction","Second Class"]
print(df)
```

```python
# Update Data in DataFrame
# Change marks of Student A to be 88
df.iloc[0,1]=88
print(df)
```

```python
# Delete Column
df1=df.drop("Grade",axis=1)
print(df)
print(df1)
```

```python
# Perfrom Basic Statistical Analysis
mean1=df["Marks"].mean()
print(mean1)
min1=df["Marks"].min()
max1=df["Marks"].max()
print(min1)
print(max1)
```

```python
# Filtering
df[df["Marks"]>80]
```
